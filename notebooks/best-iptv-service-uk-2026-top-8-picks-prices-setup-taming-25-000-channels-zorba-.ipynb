## What Viewers Want in the UK
British viewers usually judge a service on football first, then on whether the pay-TV entertainment channels and catch-up work. Catch-up matters a lot here: many households watch the evening's programmes the next day.

## Free TV in the UK
The UK has strong free TV: Freeview, BBC iPlayer, ITVX, Channel 4 and 5 cover most entertainment and news, with a TV Licence. An IPTV subscription is mainly about what those don't carry: live sport, pay-TV entertainment and international channels.

## Sport in the UK
For UK sport fans it's mostly about the Premier League (live rights split between Sky Sports and TNT Sports from 2025/26), plus the Champions League, rugby, cricket and boxing. Saturday 3 pm games aren't broadcast live in the UK, and the 17:30 kick-off is the busiest moment of the week for any streaming service.

## Is IPTV Legal in the UK?
In the UK you need a TV Licence to watch or record live TV on any service, including IPTV. Selling access to unlicensed streams has led to prison sentences here, and watching them is also unlawful.
